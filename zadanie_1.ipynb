```python
x = 10 # globalna zmienna

def zmien():
    x = 5 # nowa zmienna lokalna (nie zmienia globalnej)
    print("Lokalne x:", x)
    
zmien()
print("Globalne x:", x)
```

```python
x = 10

def zmien():
    global x # odwołanie się do globalnego x
    x = 99
    print("Lokalne x:", x)

zmien()
print("Globalne x:", x)
```
